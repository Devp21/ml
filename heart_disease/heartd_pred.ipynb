```python
df4=df3.copy()
```

```python
df4.replace({'ExerciseAngina':{'N':0,'Y':1}},inplace=True)
```

```python
df4.replace({'ST_Slope':{'Down':1,'Flat':2,'Up':3}},inplace=True)
```

```python
df4.replace({'RestingECG':{'Normal':1,'ST':2,'LVH':3}},inplace=True)
```

```python
df4.head()
```

```python
df5 = pd.get_dummies(df4,drop_first=True,dtype='float')
df5.head()
```

```python
X=df5.drop("HeartDisease",axis='columns')
y=df5.HeartDisease
X.head()
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
X_scaled[1]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X_scaled,y,test_size=0.2,random_state=30)
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
model.fit(X_train,y_train)
```

```python
model.score(X_test,y_test)
```
